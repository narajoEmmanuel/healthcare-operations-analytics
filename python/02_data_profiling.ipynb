## Data Loading

The complete 2024 dataset was validated during Data Understanding. This notebook loads the validated raw snapshot created by that workflow.

```python
import json
from pathlib import Path

import pandas as pd

raw_path = Path("../data/raw/medicare_inpatient_2024.json")

with raw_path.open(encoding="utf-8") as raw_file:
    records = json.load(raw_file)

df = pd.DataFrame(records)

expected_shape = (145_879, 15)
if df.shape != expected_shape:
    raise RuntimeError(
        f"Raw dataset shape validation failed: expected {expected_shape}, got {df.shape}."
    )

df.shape
```
